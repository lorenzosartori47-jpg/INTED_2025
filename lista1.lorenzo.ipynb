```python
def obter_dia_da_semana_condicional():
    """Pede um número de 1 a 7 e usa condições para retornar o dia."""

    print("--- Tradutor de Número para Dia da Semana ---")

    try:
        # Pede o número ao usuário
        numero = int(input("Digite um número de 1 a 7: "))

        print(f"\nO número {numero} corresponde à: ")

        if numero == 1:
            dia = "Domingo"
        elif numero == 2:
            dia = "Segunda-feira"
        elif numero == 3:
            dia = "Terça-feira"
        elif numero == 4:
            dia = "Quarta-feira"
        elif numero == 5:
            dia = "Quinta-feira"
        elif numero == 6:
            dia = "Sexta-feira"
        elif numero == 7:
            dia = "Sábado"
        else:
            dia = "❌ Erro: Número fora do intervalo (1 a 7)."

        print(f"**{dia}**.")

    except ValueError:
        print("\n❌ Erro: Entrada inválida. Por favor, digite um número inteiro.")

# Chama a função para executar
obter_dia_da_semana_condicional()
```

```python
def verificar_ano_bissexto_detalhado():
    """Pede um ano e verifica se é bissexto usando uma estrutura if/elif/else."""

    print("--- Verificador de Ano Bissexto ---")

    while True:
        try:
            # Pede a entrada do usuário
            ano = int(input("Digite um ano (ex: 2024): "))
            break
        except ValueError:
            print("🚨 Erro: Entrada inválida. Por favor, digite um número inteiro.")

    if ano % 400 == 0:
        # 1. Regra de exceção principal: Divisível por 400 (ex: 2000)
        e_bissexto = True
    elif ano % 100 == 0:
        # 2. Regra de exceção: Divisível por 100, mas não por 400 (ex: 1900)
        e_bissexto = False
    elif ano % 4 == 0:
        # 3. Regra padrão: Divisível por 4 (ex: 2024)
        e_bissexto = True
    else:
        # 4. Outros anos (ex: 2023)
        e_bissexto = False

    if e_bissexto:
        print(f"\n✅ O ano {ano} **é** bissexto.")
    else:
        print(f"\n❌ O ano {ano} **não é** bissexto.")

# Executa a função
verificar_ano_bissexto_detalhado()
```

```python
def calcular_salario_com_extras():
    """
    Calcula o salário total com adicional de 50% para horas extras (> 40h).
    Pronto para rodar no Google Colab.
    """

    print("--- Calculadora de Salário com Horas Extras ---")
    LIMITE_HORAS = 40.0
    TAXA_EXTRA = 1.5  # 50% de adicional (1.0 da hora normal + 0.5 de adicional)

    # Coleta e Validação das Horas Trabalhadas
    while True:
        try:
            horas_trabalhadas = float(input("Digite o total de horas trabalhadas na semana: "))
            if horas_trabalhadas >= 0:
                break
            else:
                print("🚨 Erro: O número de horas não pode ser negativo.")
        except ValueError:
            print("🚨 Erro: Entrada inválida. Por favor, digite um número (ex: 45.5).")

    # Coleta e Validação do Valor por Hora
    while True:
        try:
            valor_por_hora = float(input("Digite o valor da hora de trabalho (R$): "))
            if valor_por_hora >= 0:
                break
            else:
                print("🚨 Erro: O valor por hora não pode ser negativo.")
        except ValueError:
            print("🚨 Erro: Entrada inválida. Por favor, digite um número (ex: 30.50).")

    # --- Lógica de Cálculo ---

    if horas_trabalhadas <= LIMITE_HORAS:
        # Caso 1: Sem horas extras
        salario_total = horas_trabalhadas * valor_por_hora
        print(f"\n✅ Salário Total (Sem Extras): R$ **{salario_total:.2f}**")

    else:
        # Caso 2: Com horas extras

        # Salário base pelas 40 horas
        salario_base = LIMITE_HORAS * valor_por_hora

        # Cálculo das horas e valor extras
        horas_extras = horas_trabalhadas - LIMITE_HORAS
        valor_horas_extras = horas_extras * valor_por_hora * TAXA_EXTRA

        salario_total = salario_base + valor_horas_extras

        # Exibição Detalhada
        print("\n--- DETALHES DO CÁLCULO ---")
        print(f"✔️ Salário Base ({LIMITE_HORAS:.0f}h): R$ {salario_base:.2f}")
        print(f"➕ Horas Extras ({horas_extras:.2f}h): R$ {valor_horas_extras:.2f} (50% Adicional)")
        print("-" * 35)
        print(f"💰 **SALÁRIO TOTAL A PAGAR:** R$ **{salario_total:.2f}**")

# Executa a função
calcular_salario_com_extras()
```
